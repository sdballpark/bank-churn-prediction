# 📊 Exploratory Data Analysis (EDA) – Bank Churn Dataset

This notebook explores the structure, balance, and relationships in the Bank Churn dataset.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```
df = pd.read_csv("../data/raw/Churn.csv")
print("✅ Dataset Loaded — shape:", df.shape)
```

```
df.head()
```

```
df.info()
df.isnull().sum()
```

```
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)
```

```
sns.countplot(data=df, x='Churn')
plt.title('Churn Distribution')
plt.show()
```

```
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
df[num_cols].hist(bins=30, figsize=(10, 6))
plt.tight_layout()
plt.suptitle('Distribution of Numeric Features', y=1.02)
plt.show()
```

```
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Between Numeric Features')
plt.show()
```

```
cat_cols = ['gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 'InternetService', 'Contract']
for col in cat_cols:
    sns.countplot(data=df, x=col, hue='Churn')
    plt.title(f'Churn by {col}')
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```
